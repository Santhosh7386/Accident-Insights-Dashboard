```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# visualization style
sns.set(style="whitegrid")
```

```python
#Load the Dataset
df = pd.read_csv("road_accident_dataset.csv")

df.head()
```

```python
#Basic Dataset Information
#Dataset shape
df.shape
```

```python
#Column information
df.info()
```

```python
#Statistical summary
df.describe()
```

```python
#Check Missing Values
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#Distribution of Accident Severity
plt.figure(figsize=(6,4))
sns.countplot(x='Accident Severity', data=df)
plt.title("Accident Severity Distribution")
plt.show()
```

```python
#Accidents by Weather Condition
plt.figure(figsize=(8,5))
sns.countplot(y='Weather Conditions', data=df)
plt.title("Accidents by Weather Condition")
plt.show()
```

```python
#Accidents by Road Type
plt.figure(figsize=(8,5))
sns.countplot(y='Road Type', data=df)
plt.title("Accidents by Road Type")
plt.show()
```

```python
#Vehicles vs Accident Severity
plt.figure(figsize=(6,4))
sns.boxplot(x='Accident Severity', y='Number of Vehicles Involved', data=df)
plt.show()
```

```python
#Accident Trend by Time
plt.figure(figsize=(8,5))
sns.histplot(df['Time of Day'], bins=24)
plt.title("Accidents by Time of Day")
plt.show()
```

```python
#Driver Age Distribution
plt.figure(figsize=(6,4))
sns.histplot(df['Driver Age Group'], bins=30)
plt.title("Driver Age Distribution")
plt.show()
```
